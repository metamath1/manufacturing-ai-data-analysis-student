# 머신러닝: 분류 문제와 회귀 문제





## 분류 문제 (Classification)
- **정의**: 데이터를 특정 카테고리(클래스)로 분류하는 문제.
- **예시**
  - 이메일 스팸 여부 판별 (스팸/비스팸)
  - 손글씨 숫자 인식 (0~9)
  - 질병 여부 진단 (양성/음성)
- **출력**: 이산적 값 (예: `0, 1, 2` 또는 `스팸/비스팸`)
- **평가 지표**
  - 정확도 (Accuracy)
  - 정밀도 (Precision), 재현율 (Recall)
  - F1 Score, ROC-AUC





## 회귀 문제 (Regression)
- **정의**: 입력 데이터로부터 연속적인 값을 예측하는 문제.
- **예시**
  - 주택 가격 예측 (달러 단위)
  - 온도 변화 예측 (°C)
  - 주식 가격 변동 예측
- **출력**: 연속적 값 (예: `12.5`, `300,000`)
- **평가 지표**
  - 평균제곱오차 (MSE)
  - 평균절대오차 (MAE)
  - 결정계수 (R² Score)





| 구분       | 분류 문제                 | 회귀 문제             |
|------------|---------------------------|-----------------------|
| 출력       | 이산적 값                 | 연속적 값             |
| 예시       | 스팸 여부, 이미지 인식    | 가격 예측, 온도 예측  |
| 평가 지표  | 정확도, F1 Score, ROC-AUC | MSE, MAE, R² Score    |

## crosstab()

- `crosstab()`은 범주형 변수 조합별 샘플 수를 세는 데 사용

- 부서별 퇴사 여부 인원




````python
dept_attrition_count = pd.crosstab(
    emp["Department"],
    emp["Attrition"]
)

display(dept_attrition_count)
````

- 위 분석으로 ```text 부서별로 재직자는 몇 명이고, 퇴사자는 몇 명인가? ``` 즉, `Department`와 `Attrition`이라는 두 범주형 변수의 조합별 샘플 수를 구함

- 비율로 표시 `normalize="index"`를 사용

````python
dept_attrition_ratio = pd.crosstab(
    emp["Department"],
    emp["Attrition"],
    normalize="index"
)

display(dept_attrition_ratio)
````

- 이 결과는 다음과 같이 조건부 확률로 해석

```text
P(Attrition | Department)
```

- 즉, ```특정 부서에 속한 직원이 퇴사했을 확률 ```을 확인

- 야근 여부와 퇴사 여부를 확인하시오.


````python
pd.crosstab(
    emp["OverTime"],
    emp["Attrition"],
    normalize="index"
)
````

## pivot_table()

- `pivot_table()`은 범주형 변수 조합별로 특정 값의 평균, 합계, 개수 등을 집계할 때 사용


- 부서별 평균 월급

````python
dept_income_mean = pd.pivot_table(
    data=emp,
    index="Department",
    values="MonthlyIncome",
    aggfunc="mean"
)

display(dept_income_mean)
````

- 아래 그룹바이와 동일

````python
emp.groupby("Department")["MonthlyIncome"].mean()
````

- `crosstab()`이 샘플 수를 세는 데 초점이 있다면,  `pivot_table()`은 특정 컬럼의 값을 집계하는 데 초점이 있음

- 부서 × 퇴사 여부별 평균 급여

````python
dept_attrition_income = pd.pivot_table(
    data=emp,
    index="Department",
    columns="Attrition",
    values="MonthlyIncome",
    aggfunc="mean"
)

display(dept_attrition_income)
````

````python
# crosstab으로도 가능
pd.crosstab(
    emp['Department'],
    emp['Attrition'],
    values=emp['MonthlyIncome'],
    aggfunc='mean'
)
````

- 부서 × 야근 여부별 퇴사율


````python
emp_01 = emp.copy()
emp_01["Attrition_Flag"] = emp["Attrition"].map({
    "No": 0,
    "Yes": 1
})

pd.pivot_table(
    data=emp_01,
    index="Department",
    columns="OverTime",
    values="Attrition_Flag",
    aggfunc="mean"
)
````

- 부서 상관없이 일 적게하고 돈많이 받으면 퇴사 덜한다.

- 이 결과는 다음 조건부확률로 해석

```text
P(퇴사=1 | Department, OverTime)
```

즉, ```부서와 야근 여부가 주어졌을 때 퇴사할 확률```

## merge
- 데이터 병합은 여러 개의 데이터프레임을 특정 기준(열 또는 인덱스)에 따라 하나로 합치는 과정

- 관계형 데이터베이스의 'JOIN' 연산과 유사하며, 서로 다른 테이블에 분산되어 있는 관련 데이터를 통합

````python
# 고객 데이터
customer = pd.DataFrame({'cust_id' : [0,1,2,3,4,5],
                    'name' : ['철수', '영희', '길동', '영수', '수민', '동건'],
                    '나이' : [40, 20, 21, 30, 31, 18],
                    '주소' : ['대구', '서울', '부산', '대전', '제주', '인천']})

# 고객의 주문 데이터(고객 정보는 없음)
orders = pd.DataFrame({'cust_id' : [1, 1, 2, 2, 2, 3, 3, 1, 4, 9],
                    'item' : ['치약', '칫솔', '이어폰', '헤드셋', '수건', '생수', '수건', '치약', '생수', '케이스'],
                    'quantity' : [1, 2, 1, 1, 3, 2, 2, 3, 2, 1]})
````

- 그리고 이상치를 처리하는 방법도 다양한데 주로 IQR을 이용한 방법을 사용

- IQR를 이용한 방법: 사분위수 Q1과 Q3 차이의 1.5배를 넘어가는 값을 이상치Outlier로 간주


Code_Generated_Image_iqr.png

- 다음 코드를 각 컬럼별로 수행하여 진행할 수 있음

````python
# 상한과 하한을 넘어가는 이상치를 제거하는 방법

column = 'income'
q1 = emp_selected[column].quantile(0.25)
q3 = emp_selected[column].quantile(0.75)

# 1.5 * IQR
iqr = 1.5 * (q3 - q1)

# column에서 이상치를 제거
upper_cond = emp_selected[column] < (q3 + iqr)
lower_cond = emp_selected[column] > (q1 - iqr)
print(q3 + iqr, q1 - iqr)
emp_selected[ lower_cond & upper_cond ]['income'].plot(kind='box')
````

````python
# 이상치를 제거하지 않고 경계값으로 대체하는 방법 (캡핑)
emp_capped = emp_selected.copy()

upper_limit = q3 + iqr
lower_limit = q1 - iqr

# 상한/하한값으로 대체
emp_capped.loc[emp_capped['income'] > upper_limit, 'income'] = int(upper_limit)
emp_capped.loc[emp_capped['income'] < lower_limit, 'income'] = int(lower_limit)

emp_capped['income'].plot(kind='box')
````

- 한번에 한컬럼씩 처리하면 같은 코드를 계속 반복 입력해야 하므로 함수와 `for`루프에 익숙하면 다음처럼 전체를 함수로 만들어 놓고 한번에 처리 가능



````python
# 여러 컬럼 동시 처리
def remove_outliers(df, columns):
    """여러 컬럼의 이상치를 한번에 제거"""
    df_clean = df.copy()

    for column in columns:
        q1 = df_clean[column].quantile(0.25)
        q3 = df_clean[column].quantile(0.75)
        iqr = 1.5 * (q3 - q1)

        lower_cond = (df_clean[column] >= (q1 - iqr))
        upper_cond = (df_clean[column] <= (q3 + iqr))
        mask = lower_cond & upper_cond
        df_clean = df_clean[mask]

    return df_clean
````

````python
# 사용
emp_cleaned = remove_outliers(emp_selected, ['age', 'distance', 'income', 'years'])
print(f"원본: {len(emp_selected)}개 → 제거 후: {len(emp_cleaned)}개")
````

# **4. Feature Engineering**

- Feature Engineering(피처 엔지니어링)은 원본 데이터로부터 모델 학습에 유용한 새로운 특성을 생성하거나 기존 특성을 변환하는 과정.

- 대표적인 기법으로 Binning(구간화), Scaling(정규화/표준화), Encoding(인코딩) 등이 있으며, 각각 연속형 데이터의 범주화, 변수 간 스케일 통일, 범주형 데이터의 수치 변환을 수행.

- 적절한 Feature Engineering은 모델의 예측 성능을 크게 향상시킬 수 있으며, 데이터의 특성과 도메인 지식을 바탕으로 진행하는 것이 중요. 머신러닝에서 "데이터와 특성이 알고리즘보다 중요하다"는 말이 있을 정도로 핵심적인 단계.


### Binning

- Binning(구간화)은 연속형 수치 데이터를 일정한 구간으로 나누어 범주형 데이터로 변환하는 기법
- 예를 들어 나이(18~60세)를 20대, 30대, 40대, 50대로 그룹화하거나, 소득을 저소득/중소득/고소득 구간으로 나누는 것이 대표적
- 이를 통해 데이터의 노이즈를 줄이고 패턴을 명확하게 파악할 수 있으며, 이상치의 영향을 완화하고 모델의 과적합을 방지하는 효과를 얻을 수 있음

- 판다스의 `cut()` 함수를 사용하면 구간의 경계를 직접 지정할 수 있고, `qcut()` 함수는 데이터를 동일한 개수로 나누는 분위수 기반 구간화를 수행

- 간단하게 연속 수치형 변수를 범주형 변수로 만드는 방법이라 할 수 있음

- 실무에서 꽤 강력한 전처리 기법으로 사용됨

#### cut()


- 구간의 경계값을 직접 지정하여 데이터를 나누는 방식
각 구간의 데이터 개수가 불균등할 수 있음
- 예: 나이를 [0, 30, 40, 50, 60]으로 나누면 20대/30대/40대/50대 구간 생성
- 도메인 지식을 바탕으로 의미있는 구간 설정 가능


````python
# cut() 사용: 경계값을 직접 지정하여 구간화

# income을 4개 구간으로 직접 지정
emp_cleaned['income_group'] = pd.cut(emp_cleaned['income'],
                                      bins=[0, 5000, 10000, 15000, 20000],
                                      labels=['1구간', '2구간', '3구간', '4구간'])

print("income_group 분포:")
print(emp_cleaned['income_group'].value_counts().sort_index())
print()

# distance를 근거리/중거리/원거리로 구분
emp_cleaned['distance_group'] = pd.cut(emp_cleaned['distance'],
                                        bins=3,
                                        labels=['근거리', '중거리', '원거리'])

print("distance_group 분포:")
print(emp_cleaned['distance_group'].value_counts())
print()

emp_cleaned.head()
````

#### qcut()

- 분위수(quantile) 기반으로 각 구간의 데이터 개수가 동일하게 나누는 방식
- 구간의 경계값이 자동으로 결정됨
- 예: 4개 구간으로 나누면 각 구간에 25%씩 균등 배분
- 데이터의 분포가 불균형할 때 각 그룹의 샘플 수를 맞추는데 유용

````python
# qcut() 사용: 데이터를 동일한 개수로 나누기 (분위수 기반)

# income을 4개 그룹으로 균등 분할 (각 그룹 25%씩)
emp_cleaned['income_quantile'] = pd.qcut(emp_cleaned['income'],
                                          q=4,
                                          labels=['Q1', 'Q2', 'Q3', 'Q4'])

print("income_quantile 분포:")
print(emp_cleaned['income_quantile'].value_counts())
print()

# years를 3개 그룹으로 균등 분할 (신입/중견/베테랑)
emp_cleaned['years_quantile'] = pd.qcut(emp_cleaned['years'],
                                         q=3,
                                         labels=['신입', '중견', '베테랑'])

print("years_quantile 분포:")
print(emp_cleaned['years_quantile'].value_counts())
print()

emp_cleaned.head()
````

### 스케일링(Scaling)

- 서로 다른 범위(scale)를 가진 변수들을 일정한 수준으로 맞춰주는 작업
- 예: age(18), income(2000), distance(1~29) 등 변수마다 범위가 다름
- 변수 간 스케일 차이가 크면 큰 값을 가진 변수의 영향력이 과도하게 커지는 문제 발생
- 거리 기반 알고리즘(KNN, SVM)이나 경사하강법을 사용하는 모델(로지스틱회귀, 신경망)에서 필수적



#### 표준화(Standardization)

- 평균을 0, 표준편차를 1로 변환: (x - 평균) / 표준편차

$$
x_i = \frac{x_i - \mu}{\sigma}
$$

- 변환 후 데이터 범위: 대략 -3 ~ +3 (정규분포 가정 시)
- 이상치의 영향을 받음
- sklearn의 StandardScaler 사용



- 정규 분포를 평균이 0 이고 분산이 1 인  표준 정규 분포로 변환합니다.



````python
# Standardization (표준화) - pandas로 직접 구현

# 수치형 변수
numeric_cols = ['age', 'income', 'distance', 'years']

# 수치형 변수만 선택
numeric_df = emp_cleaned[numeric_cols]

# 표준화: (x - 평균) / 표준편차
standardization_df = (numeric_df - numeric_df.mean()) / numeric_df.std()

print("표준화 결과:")
print(standardization_df.head())
print()
print("표준화 후 통계:")
print(standardization_df.describe())
````

- 직접 수식을 코딩하지 않고 대부분 머신러닝 라이브러리 `sklearn`의 전처리 툴을 이용

- 사용 방법
    - 모델을 임포트, 객체를 만들고, `fit_transform()`으로 변환
    - 🔥<font color="#E95223">[주의] 지금은 전체 데이터를 한번에 변환하지만 학습, 검증 데이터로 나눠진 상태에서는 학습 데이터에 `fit_transform()`하고 검증 데이터에는 `transform()`만 해야 함.</font>

````python

# 스케일러 객체 로딩
from sklearn.preprocessing import StandardScaler

# 스케일러 만들고
std_scaler = StandardScaler()

# 스케일링 수행 후 각 자리에 집어넣기
emp_cleaned.loc[:,numeric_cols] = std_scaler.fit_transform(emp_cleaned[numeric_cols])

# 확인
emp_cleaned.describe()
````

#### 정규화(Normalization)

- 최소값을 0, 최대값을 1로 변환: (x - 최소값) / (최대값 - 최소값)

$$
x_i = \frac{x_i - \min \mathbf{x}}{\max{ \mathbf{x} }-\min{ \mathbf{x} }}
$$

- 변환 후 데이터 범위: 정확히 0 ~ 1
- 이상치의 영향을 크게 받음 (최소/최대값 기준이라서)
sklearn의 MinMaxScaler 사용



````python
# 표준화: (x - 평균) / 표준편차
normalization_df = (numeric_df - numeric_df.min()) / (numeric_df.max()-numeric_df.min())

print("정규화 결과:")
print(normalization_df.head())
print()
print("정규화 후 통계:")
print(normalization_df.describe())
````

- 머신러닝 라이브러리 `sklearn`의 전처리 툴을 이용합니다.

### 인코딩(Encording)

- 원-핫 인코딩은 간단히 말해 한 개의 요소는 True, 나머지 요소는 False로 만들어 주는 기법입니다.    

- 기계 학습 알고리즘은 숫자를 입력으로 받기 때문에 문자로 된 범주 형 데이터를 One-Hot Encording을 통해 숫자로 바꿔야 합니다.


#### 정수인코딩(레이블인코딩)

- 범주형 데이터를 정수(0, 1, 2, ...)로 변환하는 기법
- 주로 **이진 범주**(2개 값)를 0과 1로 변환할 때 사용
- 예: Male/Female → 0/1, Yes/No → 0/1
- pandas의 `map()` 또는 `replace()` 함수로 직접 매핑하여 변환할 수 있음

- 사용 범위
    - 범주가 2개인 경우 (성별, Yes/No, True/False 등)
    - 순서가 있는 범주형 데이터 (Low/Medium/High → 0/1/2)
    - 메모리 효율이 중요할 때 (One-Hot보다 컬럼 수 적음)

- 주의사항
    - 순서가 없는 다중 범주(3개 이상)에는 부적합
    - 이유: 모델이 숫자 간 크기 관계를 학습할 수 있음 (Sales=0, HR=1, R&D=2 → R&D > HR?)
    - 다중 범주는 One-Hot Encoding 사용 권장

- sklearn의 LabelEncoder
    - 정수 인코딩을 위한 sklearn 도구
    - 원래 목적은 타겟컬럼을 인코딩 하기 위해 고안
    - 하지만 인코더를 여러개 만들어 입력 컬럼에 하나씩 적용하면서 사용 가능

````python
#현재 데이터 상태 확인
emp_cleaned.head()
````

````python
# 정수 인코딩 적용

# 방법 1: map() 사용 - gender
emp_encoded = emp_cleaned.copy()

# 간단하게 판다스 map()함수를 사용
emp_encoded['gender'] = emp_cleaned['gender'].map(
    {'Male': 0, 'Female': 1}
)

# 방법 2: LabelEncoder 사용 - overtime, attrition
from sklearn.preprocessing import LabelEncoder

# LabelEncoder 객체 생성
le_overtime = LabelEncoder()
le_attrition = LabelEncoder()

# 각 컬럼에 LabelEncoder 적용
emp_encoded['overtime'] = le_overtime.fit_transform(emp_cleaned['overtime'])
emp_encoded['attrition'] = le_attrition.fit_transform(emp_cleaned['attrition'])


# 전체 결과 확인
print("정수 인코딩 완료:")
emp_encoded.head(5)
````

#### 원-핫인코딩(One-Hot Encoding)

- 범주형 데이터를 이진(0 또는 1) 벡터로 변환하는 기법
- 각 범주마다 새로운 컬럼을 생성하고, 해당 범주면 1, 아니면 0으로 표시
- 예: Department(Sales, R&D, HR) → dept_Sales, dept_R&D, dept_HR 3개 컬럼 생성
- pandas의 `get_dummies()` 또는 sklearn의 `OneHotEncoder` 사용

- 사용 범위
    - 범주가 3개 이상인 명목형(순서 없는) 데이터
    - 예: 부서(Sales, R&D, HR), 전공(Life Sciences, Medical, Marketing...), 지역 등
    - 범주 간 크기 비교가 의미 없을 때
    - 정수 인코딩의 문제점 해결
    - 예: Sales=0, R&D=1, HR=2로 인코딩하면 모델이 "HR > R&D > Sales"로 오해
    - 원-핫 인코딩은 각 범주를 독립적으로 표현하여 순서 개념 제거

- 더미 변수 트랩 (Dummy Variable Trap)
    - N개 범주를 N개 컬럼으로 만들면 다중공선성 발생
    - 해결: N-1개 컬럼만 사용 (drop_first=True 옵션)
    - 예: Sales, R&D, HR → Sales, R&D 2개만 사용 (HR은 둘 다 0이면 자동으로 HR)

````python
# 원-핫 인코딩 (One-Hot Encoding)

# 원-핫 인코딩 대상 컬럼 확인
print("다중 범주형 컬럼:")
print(f"dept: {emp_encoded['dept'].unique()}")
print(f"edu_field: {emp_encoded['edu_field'].unique()}")
print()

# get_dummies() - 기본 사용
emp_onehot = pd.get_dummies(emp_encoded, columns=['dept', 'edu_field'], dtype=int)
print(f"컬럼 수: {emp_onehot.shape[1]}개")
print()

print("데이터 미리보기:")
emp_onehot.head()
````

````python
emp_onehot_drop = pd.get_dummies(emp_encoded,
                                  columns=['dept', 'edu_field'], dtype=int,
                                  drop_first=True)

print("더미 변수 트랩 방지 (drop_first=True):")
print(f"컬럼 수: {emp_onehot_drop.shape[1]}개")
print()

print("데이터 미리보기:")
emp_onehot_drop.head()
````

    - x: X축에 표시할 변수
    - y: Y축에 표시할 변수
    - size: 점의 크기를 변수값에 따라 조정
    - alpha: 점의 투명도 (0~1, 겹침 확인에 유용)

````python
sns.scatterplot(data=emp_cleaned_orig, x='age', y='income')
plt.show()
````

#### countplot

- 범주형 데이터의 빈도를 시각화하는 차트
- 각 범주별 데이터 개수를 막대 높이로 표시
- 범주형 변수의 분포를 한눈에 파악
- 히스토그램의 범주형 버전

- 주요 파라미터
    - x 또는 y: 범주형 변수 (x는 세로 막대, y는 가로 막대)
    - hue: 그룹별로 색상 구분하여 비교
    - order: 막대의 순서 지정


````python
fig = plt.figure(figsize=(10,5))
ax = plt.axes()
sns.countplot(data=emp_cleaned_orig, x="attrition", ax=ax)
plt.show()
````

-  hue 옵션
    - 그래프를 색깔별로 구분해서 그릴때 쓰는 옵션
    - 지정되는 변수명은 범주형 변수여야 함

````python
fig = plt.figure(figsize=(10,5))
ax = plt.axes()
sns.countplot(data=emp_cleaned_orig, x="attrition", hue="by_age", ax=ax)
plt.show()

# 40, 50대로 가면 이직 비율이 떨어짐을 확인할 수 있음
````

#### heatmap

- 데이터를 색상으로 표현하는 2차원 시각화 차트
- 주로 변수 간 상관관계 행렬을 시각화할 때 사용
- 색상의 진하기로 값의 크기를 표현 (진할수록 높은 값)
- 다수의 변수 간 관계를 한눈에 파악 가능
- df.corr()로 상관관계를 계산하고 sns.headmap()에 전달

- 주요 파라미터
    - data: 시각화할 데이터 (주로 상관계수 행렬)
    - annot: 셀에 숫자 표시 여부 (True/False)
    - fmt: 숫자 표시 형식 (예: '.2f' - 소수점 2자리)
    - cmap: 색상 테마 ('coolwarm', 'viridis', 'YlGnBu' 등)
    - vmin, vmax: 색상 범위의 최소/최대값
    - cbar: 컬러바 표시 여부




````python
# 상관관계 계산
emp_cleaned_orig.corr(numeric_only=True)
````

````python
#컬럼별 상관관계를 heatmap 그래프로 그리기
sns.heatmap(emp_cleaned_orig.corr(numeric_only=True), annot=True)
plt.show()
````

#### histplot
- 연속형 데이터의 분포를 시각화하는 차트
- 데이터를 일정한 구간(bin)으로 나누고, 각 구간의 빈도를 막대로 표시
- X축: 데이터 값의 구간, Y축: 빈도(개수)

- 주요 파라미터
    - bins: 구간의 개수 (기본값: 10, 많을수록 세밀함)
    - kde: 확률 밀도 함수 곡선 추가 (True/False)
    - hue: 그룹별로 색상 구분

````python
sns.histplot(data=emp_cleaned_orig, x='age', bins=20, kde=True)
````

#### kdeplot

````python
sns.kdeplot(data=emp_cleaned_orig, x='age')
````

#### boxplot

- 데이터의 분포와 이상치를 시각화하는 차트
- 5가지 주요 통계값을 박스와 선으로 표현
- 박스: Q1(25%), Q2(50%, 중앙값), Q3(75%)
- 수염(whisker): Q1 - 1.5×IQR ~ Q3 + 1.5×IQR
- 점: 이상치 (수염 범위 밖의 데이터)
- 여러 그룹 간 분포 비교에 유용

- 주요 파라미터
    - x: 범주형 변수 (그룹)
    - y: 연속형 변수 (분포를 확인할 값)
    - data: 데이터프레임
    - hue: 그룹 내 세부 분류
    - order: 박스의 순서 지정
    - orient: 박스 방향 ('v': 세로, 'h': 가로)



````python
sns.boxplot(data=emp_cleaned_orig, y="income", x="by_age", hue="gender", width=0.9)
````

#### barplot

- 범주에 따른 수치값의 평균을 막대 그래표로 시각화하는 차트
- 여러 그룹 간 수치값 비교에 유용

- 주요 파라미터
    - x: 범주형 변수 (그룹)
    - y: 연속형 변수 (분포를 확인할 값)
    - data: 데이터프레임
    - hue: 그룹 내 세부 분류

````python
sns.barplot(data=emp_cleaned_orig, y="income", x="by_age", hue='gender', width=0.9)
````
